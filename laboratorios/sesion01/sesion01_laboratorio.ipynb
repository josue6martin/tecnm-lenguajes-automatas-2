### Fase 2: El Científico (Prompting C-R-E-O)
Pega aquí la implementación optimizada propuesta por la IA:

```python
%%writefile ia_sesion01_c.c
#include <stdio.h>
#include <string.h>

// Implementación optimizada aplicando contrato C-R-E-O (O(n) tiempo, O(1) memoria)
void invertir_buffer_ia(char *buffer, int len) {
    if (buffer == NULL || len <= 0) return;

    char *left = buffer;
    char *right = buffer + len - 1;

    while (left < right) {
        char temp = *left;
        *left = *right;
        *right = temp;
        left++;
        right--;
    }
}
```

### Fase 3: El Auditor (Test Harness en C)
Pruebas de estrés: Valores NULL, buffers vacios y longitudes impares

```python
%%writefile harness_sesion01_c.c
#include <stdio.h>
#include <assert.h>
#include <string.h>

// Declaración de la función optimizada generada en la Fase 2
void invertir_buffer_ia(char *buffer, int len);

int main() {
    printf("=== AUDITORIA CRITICA: SESION 01 ===\n");

    // Prueba 1: Buffer NULL (Verifica que no falle / no genere segmentation fault)
    invertir_buffer_ia(NULL, 0);
    printf("Prueba NULL: PASO ✅\n");

    // Prueba 2: Buffer vacío
    char vacio[] = "";
    invertir_buffer_ia(vacio, 0);
    assert(strcmp(vacio, "") == 0);
    printf("Prueba Buffer Vacío: PASO ✅\n");

    // Prueba 3: Longitud impar ("abcde" -> "edcba")
    char impar[] = "abcde";
    invertir_buffer_ia(impar, 5);
    assert(strcmp(impar, "edcba") == 0);
    printf("Prueba Longitud Impar ('abcde'): PASO ✅\n");

    // Prueba 4: Longitud par ("abcd" -> "dcba")
    char par[] = "abcd";
    invertir_buffer_ia(par, 4);
    assert(strcmp(par, "dcba") == 0);
    printf("Prueba Longitud Par ('abcd'): PASO ✅\n");

    printf("Ejecutando pruebas de estrés... PASO ✅\n");
    return 0;
}
```
